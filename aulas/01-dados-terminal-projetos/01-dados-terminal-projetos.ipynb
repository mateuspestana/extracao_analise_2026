## Isso já tem nome

O que você acabou de perceber é uma distinção real do trabalho com dados, que vimos em Introdução à Ciência de Dados, no semestre passado:

- **Estruturado** — todas as linhas seguem as mesmas colunas, como `publicacoes.csv`. Ótimo para filtrar, contar e comparar.
- **Semiestruturado** — há organização por chaves e listas, mas nem todo registro precisa ter o mesmo formato, como `perfil.json`.
- **Desestruturado** — conteúdo rico, mas sem linhas e colunas prontas, como `anotacoes.txt` e, de um jeito diferente, `pagina.html` (que é estruturado por marcação, não por dado).

Nenhum formato é melhor que o outro, a pergunta da pesquisa é que decide o que fazer com cada um.

## Voltando ao Observatório: fonte, dado e informação

A conta do Observatório Local é a **fonte**. Os quatro arquivos que você acabou de abrir são os **dados**. E uma frase como "as publicações sobre participação tiveram menos curtidas nesta amostra" é **informação**, uma leitura produzida a partir dos dados, não o dado em si.

Toda vez que você encontrar uma fonte nova, vale perguntar: quem publicou? quando foi atualizada? o que cada registro representa? e, principalmente, **o que essa fonte não mede?** Essas perguntas evitam tirar conclusões maiores do que os dados permitem.

## Antes de confiar num arquivo: espiar e registrar

Um arquivo chamado `dados.csv` pode separar colunas por vírgula, ponto e vírgula ou tabulação; um `relatorio-final.pdf` pode conter uma tabela sem ser uma planilha editável. Por isso, antes de escrever qualquer código, vale abrir o arquivo e olhar algumas linhas.

Quando a fonte for de verdade, registre também: nome, link, data de acesso, período coberto, formato e uma limitação. Parece burocracia, mas é o que evita a pergunta "de onde saiu esse número?" dias depois.

Exemplo: *"Publicações da conta do Observatório Local; acesso em 29/07/2026; posts entre 20 e 22 de julho; CSV exportado; não informa quantas pessoas visualizaram cada post."*

## Precisamos de um jeito de ir e vir

Você já sabe ler os arquivos do Observatório. Agora, para organizar essa investigação (e todas as que vêm pela frente no semestre) vamos precisar entrar e sair de pastas sem depender de clique duplo. É para isso que serve o terminal.

Dois lembretes rápidos antes de abrir:

- Um **arquivo** guarda conteúdo (uma foto, uma planilha, um texto, um programa). Uma **pasta** organiza arquivos e outras pastas. Um **caminho** é o endereço de uma pasta ou arquivo no computador, no Windows costuma usar barra invertida (`C:\Users\matheus.pestana\Documents`), no Mac usa barra para frente (`/Users/matheus.pestana/Documents`).
- Você não precisa decorar caminho nenhum. O terminal deixa caminhar uma pasta de cada vez. O hábito que interessa é perguntar sempre: onde estou? o que existe aqui? para onde quero ir agora?

## Abrindo o terminal

Abra o Prompt de Comando (CMD) ou o Terminal agora. Antes de rodar o comando abaixo, palpite: o que você acha que vai aparecer, um nome de pasta, uma lista de arquivos, algo em branco?

**Windows (CMD)** — em qual pasta estou agora?
```cmd

```

**Mac (Terminal)**
```bash

```

Confira: apareceu um caminho. É o endereço da pasta onde o terminal está "parado" agora. Em seguida, peça para listar o que existe aqui dentro.

**Windows:** `dir`
**Mac:** `ls`

## Mini prática: ande por uma pasta que você conhece

Antes de criar qualquer coisa nova, pratique indo e vindo em uma pasta qualquer do seu computador:

1. Rode `dir` (Windows) ou `ls` (Mac) para listar o que tem aqui.
2. Escolha um nome da lista e entre com `cd nome-da-pasta`.
3. Liste de novo, para conferir que mudou de lugar.
4. Volte com `cd ..`.

Truque que evita metade dos erros de digitação: aperte **Tab** para o terminal completar um nome de arquivo ou pasta sozinho.

**E se der erro?** Se aparecer algo como `O sistema não pode encontrar o caminho especificado` (Windows) ou `no such file or directory` (Mac), CALMA!!! O terminal está te dizendo exatamente qual nome ele não reconheceu. Releia o comando, confira maiúsculas, espaços e acentos, e tente de novo. Ler a mensagem de erro com calma é uma habilidade que vamos treinar a aula inteira.
