# Aula 3 — Git, GitHub e repositório profissional

Nesta aula, vamos registrar mudanças no projeto e enviá-las para um repositório privado no GitHub. O objetivo é que outra pessoa consiga abrir o repositório, entender o que existe nele e encontrar cada entrega. Pense no Git como o histórico bem organizado do seu trabalho: ele evita que versões importantes virem um mistério.

## 1. O que cada ferramenta faz

**Git** é o programa que registra versões dos arquivos no seu computador. **GitHub** é o serviço que hospeda uma cópia remota do repositório. Um **commit** é um registro de uma mudança acompanhada de uma mensagem.

Git não salva tudo automaticamente. Você primeiro confere o que mudou, escolhe o que será incluído e então cria um commit. Essa pausa para conferir é uma boa prática, não uma perda de tempo.

## 2. Preparação

Abra o Prompt de Comando (CMD) no Windows ou Terminal no Mac e confira se Git está instalado:

```cmd
git --version
```

Se for a primeira vez que usa Git nesse computador, informe seu nome e e-mail. Eles aparecem como autoria dos commits; use dados que você reconhece.

```cmd
git config --global user.name "Seu Nome"
git config --global user.email "seu-email@exemplo.com"
```

Os mesmos comandos funcionam no CMD e no Mac.

## 3. Criando o repositório privado no GitHub

1. Entre no GitHub com sua conta.
2. Clique no sinal `+` no canto superior direito e escolha **New repository**.
3. Dê um nome claro, por exemplo `extracao-dados-trabalhos-seunome` (COLOQUE AQUI O SEU NOME! NÃO O MEU!!!)
4. Selecione **Private**.
5. Não marque a opção para criar README neste momento: vamos enviar o README que já está no computador.
6. Clique em **Create repository**.
7. Deixe a página aberta: ela mostra a URL que será usada mais adiante.

O repositório é privado porque as entregas são individuais. Compartilhe acesso apenas quando eu solicitar.

## 4. Preparando os arquivos locais

Abra no VS Code a pasta `extracao-dados-trabalhos-SeuNome` criada na Aula 1. Se não a criou, volte na primeira aula para fazê-lo. Crie nela um arquivo `README.md`, um arquivo `.gitignore` e a pasta `projetos/`. Dentro de `projetos/`, crie as pastas `01-ambiente/` e `02-python-basico/`.

No CMD, a estrutura pode começar assim:

```cmd
mkdir projetos
mkdir projetos\01-ambiente
mkdir projetos\02-python-basico
```

No Mac, use barras para frente: `mkdir -p projetos/01-ambiente projetos/02-python-basico`. Copie a entrega da Aula 1 para `projetos/01-ambiente/`.

No `.gitignore`, escreva estas linhas:

```text
.venv/
.env
__pycache__/
*.py[cod]
.DS_Store
Thumbs.db
```

O arquivo `.gitignore` lista itens que **não** devem entrar no Git. A pasta `.venv` pode ser recriada; arquivos `.env` podem guardar segredos. Código, documentação e `requirements.txt`, por outro lado, normalmente devem ser versionados.

## 5. Primeiro commit, em etapas

No terminal aberto na pasta do seu repositório, execute uma linha por vez:

```bash
git init
git status
git add README.md .gitignore projetos
git status
git commit -m "cria estrutura inicial dos trabalhos"
```

O primeiro `git status` mostra arquivos que Git ainda não acompanha. `git add` prepara apenas os arquivos listados. O segundo `git status` permite conferir a seleção antes do commit.

Se você vir `.env`, `.venv/`, um arquivo grande inesperado ou algo que não reconhece, não faça o commit ainda. Revise o `.gitignore` e peça ajuda. Não use comandos de remoção por impulso.

## 6. Conectando ao GitHub e enviando

Na página do repositório recém-criado, copie a URL HTTPS. Ela se parece com `https://github.com/seu-usuario/extracao-dados-trabalhos-seunome.git`. Substitua a parte entre sinais de menor e maior pela URL real:

```bash
git branch -M main
git remote add origin <URL_DO_REPOSITORIO>
git push -u origin main
```

Depois abra a página do GitHub e atualize-a. Você deve ver o README, `.gitignore` e a pasta `projetos/`. A primeira vez pode abrir uma janela pedindo autenticação; siga o fluxo do GitHub no navegador.

**OBS:** o GitHub não aceita mais senha direto no terminal. Se aparecer um pedido de usuário/senha em vez do fluxo do navegador, entre com seu usuário e cole um *personal access token* no lugar da senha. Podemos gerar esse token juntos se precisar.

## 7. A rotina de cada entrega

Quando modificar uma atividade, use esta sequência:

```bash
git status
git add nome-do-arquivo-ou-pasta
git commit -m "descreve a mudança feita"
git push
```

Mensagens boas descrevem a ação: `adiciona leitura inicial de csv` ou `documenta fonte dos dados`. Mensagens como `update` ou `teste` não ajudam você a lembrar o que foi feito. Faça pelo menos um commit de progresso antes do commit de entrega; o histórico deve mostrar o desenvolvimento do trabalho.

## 8. E quando alguém atualiza o repositório?

No repositório de materiais do curso, antes de uma aula, use:

```bash
git pull
```

Se Git informar um conflito, calma!!! Ele só está dizendo que duas versões alteraram a mesma parte de um arquivo. A forma mais rápida de resolver é:

```bash
git reset --hard HEAD
```

Isso faz com que tudo seja zerado e fique de acordo com o repositório original da disciplina. 

Agora, se isso aconteceu com o SEU repositório, não use o comando, ou as mudanças serão perdidas. Avalie onde está a diferença, e então faça o commit da versão correta.
