Abra `data/publicacoes_exemplo.csv` no VS Code. Leia a primeira linha: ela é o **cabeçalho** e dá nome às colunas. Cada linha abaixo é uma publicação fictícia.

Antes de usar código, responda mentalmente: quantas colunas existem? Qual coluna identifica cada publicação? Quais campos parecem contar interações? Essa olhada inicial evita que a gente trate um arquivo como uma caixa-preta.

## 8. Ler o arquivo, sem atalhos

A célula abaixo abre o CSV diretamente. `open(...)` recebe o nome do arquivo, `csv.reader` lê as linhas e `next` pega uma linha de cada vez. Execute-a e compare o resultado com o arquivo aberto no editor.

````python
import csv

with open("data/publicacoes_exemplo.csv", encoding="utf-8", newline="") as arquivo:
    leitor = csv.reader(arquivo)
    cabecalho = next(leitor)
    primeira_linha = next(leitor)

print("Cabeçalho:", cabecalho)
print("Primeira publicação:", primeira_linha)
````

A primeira publicação aparece como uma lista. Isso funciona, mas ainda exige lembrar que o quarto valor representa curtidas, por exemplo. Para trabalhar com nomes de coluna, usamos `csv.DictReader`.

````python
with open("data/publicacoes_exemplo.csv", encoding="utf-8", newline="") as arquivo:
    leitor = csv.DictReader(arquivo)
    linhas = list(leitor)
    colunas = leitor.fieldnames or []

print(f"Linhas de dados: {len(linhas)}")
print(f"Colunas: {', '.join(colunas)}")
print("Tema da primeira publicação:", linhas[0]["tema"])
````

Agora cada linha tem nomes: `linhas[0]["tema"]` procura o valor da coluna `tema` na primeira publicação. O resultado esperado é 12 linhas e cinco colunas.

Repare em um detalhe: o CSV chega como texto. Para somar curtidas, primeiro precisamos transformar o valor em número.

````python
curtidas_da_primeira_publicacao = int(linhas[0]["curtidas"])
comentarios_da_primeira_publicacao = int(linhas[0]["comentarios"])
interacoes_da_primeira_publicacao = curtidas_da_primeira_publicacao + comentarios_da_primeira_publicacao

print(f"Interações da primeira publicação: {interacoes_da_primeira_publicacao}")
````

## 9. Quando der errado

Erros fazem parte do trabalho: não leia só a última linha, procure duas pistas, o **nome do erro** e a **linha indicada**.

- `FileNotFoundError`: o Python não encontrou o arquivo. Confira o nome e se a pasta certa está aberta.
- `SyntaxError`: uma regra de escrita foi quebrada, como aspa ou parêntese faltando.
- `KeyError`: você pediu uma coluna que não existe, ou escreveu o nome diferente do cabeçalho.

Não chute três mudanças ao mesmo tempo: leia, teste uma hipótese pequena e execute de novo.
